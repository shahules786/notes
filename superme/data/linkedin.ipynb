```python
from linkedin_api import Linkedin

# Authenticate using any Linkedin user account credentials
api = Linkedin('', '')


# GET a profile
profile = api.get_profile('shahules')
posts = api.get_profile_posts('shahules',post_count=100)
```

```python
keys = ['skills','experience','education','volunteer','certifications','languages','projects','publications','honors']
data = api.get_profile('shahules')
```

```python
profile_details = {}
import json
for key in keys:
    profile_details.update({key:data[key]})
    
with open("shahules_profile.json", 'w', encoding='utf-8') as f:
    json.dump(profile_details, f, ensure_ascii=False, indent=2)
```

```python
import json
import re

# Create an empty dictionary to store the transformed posts
transformed_posts = {}

# Loop through each post in the posts list
for post in posts:
    try:
        # Extract post ID from the urn in updateMetadata
        # Format: "urn:li:activity:7302526679577505792"
        post_id = post['updateMetadata']['urn'].split(':')[-1]
        
        # Extract content from commentary.text.text
        content = post['commentary']['text']['text'] if 'commentary' in post and 'text' in post['commentary'] else ""
        
        # Extract URL from the SHARE_VIA action in updateMetadata.updateActions.actions
        url = None
        for action in post['updateMetadata']['updateActions']['actions']:
            if action['actionType'] == 'SHARE_VIA':
                url = action['url']
                break
        
        # If URL not found in actions, try socialContent.shareUrl
        if not url and 'socialContent' in post and 'shareUrl' in post['socialContent']:
            url = post['socialContent']['shareUrl']
        
        # Extract date from actor.subDescription.text
        # Format: "1mo •   "
        date_text = post['actor']['subDescription']['text']
        date = re.match(r'(\d+\w+)', date_text).group(1) if re.match(r'(\d+\w+)', date_text) else date_text
        
        # Add the transformed post to the dictionary
        transformed_posts[post_id] = {
            "content": content,
            "url": url,
            "date": date
        }
    
    except Exception as e:
        print(f"Error processing post with error: {e}")
        continue

# Save the transformed posts to a JSON file
with open("shahules_posts.json", 'w', encoding='utf-8') as f:
    json.dump(transformed_posts, f, ensure_ascii=False, indent=2)

print(f"Successfully saved {len(transformed_posts)} posts to shahules_posts.json")

# Display the transformed data
```
